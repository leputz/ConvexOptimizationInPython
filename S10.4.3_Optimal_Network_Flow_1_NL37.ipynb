# Demo: Further Readings on Optimal Network Flow


Consider a directed graph with $n$ edges and $m$ nodes, $x_j$ is the flow on arc $j$. The \textit{node-arc incidence matrix}, $A\in R^{m\times n}$ of the graph is defined as

\begin{equation}
A_{i,j}=\left\{
\begin{array}{rccccccc}
1 & arc & j &  is & leaving & from & node &  i \\
-1 & arc & j & is & entering & into & node &  i \\
0 & {otherwise}
\end{array} \right.
\end{equation}

The cost function on arc $j$ with a traffic flow $x_j$ is defined as: $\phi_j(x_j):\mathbf{R}\rightarrow\mathbf{R}$. One form of the optimal network problem is to choose the best flow vector $x$ while satisfying the flow balance equation, $Ax=b$. 
\begin{equation}
\begin{array}{rc}
\text{minimize} & \displaystyle{\sum_{i=1}^n}\phi_i(x_i) \\
\text{subject to} & Ax=b \\
\end{array}
\end{equation}

Here $A$ may not be full ranked if $1^Tb=0$, i.e., the total flow within the network is conserved. In practice, further bounds on $x$ may be applied, i.e., $\{l \leq x \leq u\}$. If the cost function is linear, $\displaystyle{\sum_{i=1}^n}\phi_i(x_i)=c^Tx$, it becomes a minimum cost network flow problem: 
\begin{equation}
\begin{array}{rc}
\text{minimize} & c^Tx\\
\text{subject to} & Ax=b \\
\end{array}
\end{equation}

Another special case is the maximum flow problem (left), which is also equivalent to a minimum cost flow problem (right).
\begin{equation}
\begin{array}{rlcrl}
\text{maximize} & t & & \text{minimize} & -t \\
\text{subject to} & Ax=te & \Longleftrightarrow & \text{subject to} & \left[A -e\right]\left[\begin{array}{c}
x \\
t \\
\end{array}\right] = 0\\
& l \leq x \leq u & & & l \leq x \leq u\\
& e=(1,0,...,0,-1)& & & e=(1,0,...,0,-1) \\
\end{array}
\end{equation}

Other special examples include the minimum cut problem solved by Ford-Fulkerson Algorithm,  the Min-Cost Max-Flow problem.


It is noted $A$ is a \textbf{totally unimodular matrix}, because all of its minors are +1, -1 or 0 where a minor is the determinant of a square sub-matrix. As a totally unimodular matrix, it can be proved that:

# The inverse of any non-singular square sub-matrix of a totally unimodular matrix has entries +1, -1 or 0.

Furthermore, the integrality of extreme points can also be guaranteed for the polyhedron $P=\{x|Ax = b, l\leq x \leq u\}$ in $\mathbf{R}^n$ where $b, l, u$ are integer vectors, and $A$ is a totally unimodular matrix.
 
# All the extreme points of $P$ are integer vectors using rank test on polyhedron $\{x|\tilde{A}x\leq \tilde{b}\}$, where $\tilde{A}=\left[\begin{array}{c}
A \\
I \\
I \\
\end{array}\right]$, and $\tilde{b}=\left[\begin{array}{c}
b \\
-l \\
u \\
\end{array}\right]$


Such integrality can be used to transform some difficult integer LP to its relaxation as general LP when $A$ is a totally unimodular matrix, and $b,l,u$ are integers. Some examples include,
\begin{itemize}
\item Combinatorial optimization vs. LP relaxation 
$$
\begin{array}{rlcrl}
\text{minimize} & c^Tx & & \text{minimize} & c^Tx \\
\text{subject to} & Ax=b & \Longleftrightarrow & \text{subject to} & Ax=b \\
& l \leq x \leq u, x\in \mathbf{Z}^n & & & l \leq x \leq u\\
\end{array}
$$
\item Shortest path problem vs. LP relaxation (whose optimal solutions are extreme points satisfying $x_i\in\{0,1\}$
$$
\begin{array}{rlcrl}
\text{minimize} & 1^Tx & & \text{minimize} & 1^Tx \\
\text{subject to} & Ax=(1,0,...,0,-1) & \text{vs.} & \text{subject to} & Ax=(1,0,...,0,-1) \\
& x\in\{0,1\}^n & & & 0\leq x \leq 1\\
\end{array}
$$
\item Weighted bipartite matching problem: (1) match $N$ persons to $N$ tasks; (2) each person assigned to one task; (3) each task assigned to one person; (4) the cost of matching person $i$ to $j$ is $A_{i,j}$. 
\begin{equation}
\begin{array}{rl}
\text{minimize} & \displaystyle{\sum_{i,j=1}^N}A_{i,j}X_{i,j} \\
\text{subject to} & \displaystyle{\sum_{i=1}^N}X_{i,j}=1, j=1,...,N \\
& \displaystyle{\sum_{j=1}^N}X_{i,j}=1, i=1,...,N \\
& X_{i,j}\in \{0,1\},i,j=1,...,N \\
\end{array}
\end{equation}
Its LP relaxation is
\begin{equation}
\label{eq:NL37-wbm-lp}
\begin{array}{rl}
\text{minimize} & \displaystyle{\sum_{i,j=1}^N}A_{i,j}X_{i,j} \\
\text{subject to} & \displaystyle{\sum_{i=1}^N}X_{i,j}=1, j=1,...,N \\
& \displaystyle{\sum_{j=1}^N}X_{i,j}=1, i=1,...,N \\
& 0 \leq X_{i,j} \leq 1, i,j=1,...,N \\
\end{array}
\end{equation}
It is proved by the Birkhoff theorem that, $X$ is a doubly stochastic matrix in $\mathbf{R}^{N\times N}$, and the extreme points of the polyhedron $\hat{P}=\{X|\hat{A}X=1_{2N}\}$ have entities $\in \{0,1\}$ (actually the optimal solutions are permutation matrices). As a result, the optimal solutions from the LP relaxation are also the solutions of the weighted bipartite matching problem. Here, 
$$
\hat{A}=\left[\begin{array}{c}
\mathbf{diag}_N(\mathbf{1}_N^T, ..., \mathbf{1}_N^T) \\
I_N  \odot \mathbf{1}^T_N \\
\end{array} \right]
$$
is a totally unimodular matrix. Note that though $\hat{A}$ is not full ranked; however, its reduced form by removing any one of its row is full ranked with the rank equal to $2N-1$.
\end{itemize}

In this demo, the weighted bipartite matching problem with $N=20$ and $N=30$ is solved in its LP relaxation form using the \textit{scipy.optimize.linprog}. In the experiment, $A_{i,j}$ is randomly generated in $\mathbf{U}(0,1)$.









```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import FormatStrFormatter
from scipy.optimize import linprog
```

The weighted bipartitle matching problem is
\begin{equation}
\begin{array}{ccccc}
minimize & & \sum_{i,j=1}^N A_{i,j}X_{i,j} \\
subject & to & \sum_{i=1}^N X_{i,j}=1,j=1,...,N \\
& & \sum_{j=1}^N X_{i,j}=1,i=1,...,N \\
& & X_{i,j}\in \{0, 1\}, i,j=1,...,N \\
\end{array}
\end{equation}

Its relaxation is acquired by changing 

\begin{equation}
X_{i,j}\in \{0, 1\}, i,j=1,...,N
\end{equation}

to 

\begin{equation}
0\leq X_{i,j}\leq 1, i,j=1,...,N
\end{equation}

After such relaxation, it is a LP programming problem and can be rewritten in the general minimization problem

\begin{equation}
\begin{array}{ccc}
minimize & & c^T x \\
subject & to & Ax=b \\ 
& & 0\leq x \leq 1 \\
\end{array}
\end{equation}

This relaxation problem can be solved by scipy.optimize.linprog directly.

```python
def solve_weighted_bipartite_matching_problem_by_its_LP_relaxation(N, ax0, ax1, seed):
    """
    The weighted bipartitle matching problem is
        minimize $\sum_{i,j=1}^N A_{i,j}X_{i,j}$
        subject to $\sum_{i=1}^N X_{i,j}=1,j=1,...,N$
                   $\sum_{j=1}^N X_{i,j}=1,i=1,...,N$
                   $X_{i,j}\in \{0, 1\}, i,j=1,...,N$

    Its relaxation is acquired by changing $X_{i,j}\in \{0, 1\}, i,j=1,...,N$ to $0\leq X_{i,j}\leq 1, i,j=1,...,N$
    After such relaxation, it is a LP programming problem and can be rewritten in the general minimization problem
        minimize $c^T x$
        subject to $Ax=b$
                   $0\leq x \leq 1$
    This relaxation problem can be solved by scipy.optimize.linprog directly.
    :param N:
    :param ax:
    :param seed:
    :return:
    """
    N2 = N * N
    np.random.seed(2)

    C = np.random.rand(N, N)
    C_v = np.reshape(C, (-1, 1))
    c = np.squeeze(C_v)
    I_N = np.identity(N)
    A1 = np.tile(I_N, N)
    A2 = np.repeat(I_N, N, axis=1)
    A1A2 = np.block([[A1], [A2]])
    A = A1A2[0:-1, :]
    if np.linalg.matrix_rank(A) != 2 * N - 1:
        raise RuntimeError('rank(A) != 2N-1?')

    b = np.ones((2 * N - 1, 1))

    Aeq = A
    beq = np.squeeze(b)
    bound = (0, 1)
    bounds = [bound for _ in range(N2)]
    res = linprog(c, A_ub=None, b_ub=None, A_eq=Aeq, b_eq=beq, bounds=bounds)

    Z = np.reshape(res.x, (N, N))
    ax0.pcolor(Z, edgecolors='k', cmap='RdPu')
    ax0.set_xlabel(r'$X_{:,j}$')
    ax0.set_ylabel(r'$X_{i,:}$')
    ax0.set_title(r'optimal $X^*$ when N={}'.format(N))
    ax0.xaxis.set_major_formatter(FormatStrFormatter('%d'))
    ax0.yaxis.set_major_formatter(FormatStrFormatter('%d'))

    ax1.plot(res.x, marker='+', linestyle='none')
    ax1.set_xlabel('flatten(X)')
    ax1.grid()
```

```python
if __name__ == "__main__":

    Ns = [20, 30]
    plt.figure(figsize=(8, 8))
    for N_i, N in enumerate(Ns):

        ax0 = plt.subplot(2, 2, N_i * 2 + 1)
        ax1 = plt.subplot(2, 2, N_i * 2 + 2)
        solve_weighted_bipartite_matching_problem_by_its_LP_relaxation(N=N, ax0=ax0, ax1=ax1, seed=1)

    plt.tight_layout()
    plt.show()
```
